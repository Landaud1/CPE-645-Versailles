# Strategy:
1. Identify area of focus
2. Create a histogram to identify the intensities of the stains
3. Create a mask to look at the stains (experiment with threshold)
3. Write a piecewise thresholding function to create a floor for intensities to replace the stains with the mean of the background

```python
from PIL import Image
img = Image.open("FRAN_0124_3365_L.jpg")
display(img)
```

```python
region = img.crop((100,4000,1000,6000))
display(region)
```

```python
import numpy as np
import matplotlib.pyplot as plt

gray = np.array(region.convert("L"))

plt.hist(gray.flatten(), bins=256, range=(0,256))
plt.xlabel("Pixel intensity (0 = black, 255 = white)")
plt.ylabel("Number of pixels")
plt.title("Versailles paper intensity histogram")
[plt.show]
```

```python
dark_pixels = gray < 210
plt.figure(figsize=(8,8))
plt.imshow(~dark_pixels, cmap="gray", vmin=0, vmax=1)
plt.title("Pixels with intensity below 210")
plt.axis("off")
plt.show
```

```python
threshold = 210
selected = gray < threshold
percentage = 100 * np.mean(selected)
print(f"Pixels below {threshold}: {percentage: .2f}%")
```

```python
rgb = np.array(region.convert("RGB"))
background = np.median(rgb[gray >= 210], axis=0).astype(np.uint8)
print("Estimated background RGB:", background)
```

```python
restored = rgb.copy()
restored[gray < 210] = background
fig, axes = plt.subplots(1, 2, figsize=(12, 8))
axes[0].imshow(rgb)
axes[0].set_title("Original")
axes[1].imshow(restored)
axes[1].set_title("Restored - threshold 210")
for ax in axes:
    ax.axis("off")
plt.show()
```
