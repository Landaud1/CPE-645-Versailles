# Strategy:
1. Identify areas of focus
2. Analyze the noise
3. Apply noise reduction filters (median vs gaussian)
4. Experiment with filter parameters, similar to HW2

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from PIL import Image
from google.colab import drive

img = Image.open("/content/drive/MyDrive/CPE 645 Image Processing and Computer Vision/Project/Versailles_FP/data1/Test/Frames/frame/FRAN_0124_3365_L.jpg")

w, h = img.size

x1, y1 = int(w * 0.20), int(h * 0.58)
x2, y2 = int(w * 0.50), int(h * 0.82)

crop_c3 = img.crop((x1, y1, x2, y2))

fig, ax = plt.subplots(1, 2, figsize=(15, 6))

ax[0].imshow(img)
ax[0].add_patch(Rectangle(
    (x1, y1), x2-x1, y2-y1,
    fill=False, edgecolor="red", linewidth=2
))
ax[0].set_title("Original - Selected Region")
ax[0].axis("off")

ax[1].imshow(crop_c3)
ax[1].set_title("C3 - Area of Focus")
ax[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
from PIL import ImageOps, ImageEnhance
import numpy as np

gray_c3 = ImageOps.grayscale(crop_c3)
enhanced_c3 = ImageEnhance.Contrast(gray_c3).enhance(3)

fig, ax = plt.subplots(1, 3, figsize=(16,5))

ax[0].imshow(gray_c3, cmap="gray", vmin=0, vmax=255)
ax[0].set_title("Original Grayscale")

ax[1].imshow(enhanced_c3, cmap="gray", vmin=0, vmax=255)
ax[1].set_title("Contrast Enhanced (For Inspection)")

ax[2].hist(np.array(gray_c3).ravel(), bins=64, range=(0,255))
ax[2].set_title("Pixel Intensity Histogram")
ax[2].set_xlabel("Intensity (0-255)")
ax[2].set_ylabel("Pixel Count")

for a in ax[:2]:
  a.axis("off")

plt.tight_layout()
plt.show()
```

```python
from PIL import ImageFilter

median_c3 = crop_c3.filter(ImageFilter.MedianFilter(size=5))
gaussian_c3 = crop_c3.filter(ImageFilter.GaussianBlur(radius=3))

fig, ax = plt.subplots(1, 3, figsize=(18, 6))

images = [crop_c3, median_c3, gaussian_c3]
titles = [
    "Original",
    "Median Filter (5x5)",
    "Gaussian Filter (radius=3)"
]

for a, image, title in zip(ax, images, titles):
  a.imshow(image)
  a.set_title(title)
  a.axis("off")

plt.tight_layout()
plt.show()
```
