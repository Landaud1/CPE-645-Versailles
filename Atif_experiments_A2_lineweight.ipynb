# A2 Lineweight Variation using Image Example 1
### Note: this is experiment is for lineweight variation not lineweight desaturation, which is Experiment B2.
# Strategy:
1. Identify area of focus
2. Convert to grayscale
3. Identify line to darken by variation in its lineweight intensity along a single run (note for the experiment, we are separating this from line thickness, which also contributes to lineweight)
4. Allocate darker intensities to the identified line using intensity level slicing

```python
from PIL import Image
from google.colab import drive

# Mount Google Drive to access your files
drive.mount('/content/drive')

# Open and display the image
img = Image.open("/content/drive/MyDrive/CPE 645 Image Processing and Computer Vision/Project/Versailles_FP/data1/Test/Frames/frame/FRAN_0124_3365_L.jpg")
display(img)
```

```python
region = img.crop((3400,5160,4600,5200))
display(region)
```

```python
import numpy as np
import matplotlib.pyplot as plt

gray = np.array(region.convert("L"))

plt.figure(figsize=(14,3))
plt.imshow(gray, cmap="gray", vmin=0, vmax=255)
plt.title("A2 - Grayscale Wall Outline")
plt.axis("off")
plt.show()
```

```python
left = gray[:,0:400]
center = gray[:, 400:800]
right = gray[:, 800:1200]

print("Mean grayscale intensity:")
print(f"Left: {left.mean():.2f}")
print(f"Center: {center.mean():.2f}")
print(f"Right: {right.mean():.2f}")
```

```python
row_means = gray.mean(axis=1)

plt.figure(figsize=(10,4))
plt.plot(range(len(row_means)), row_means)

plt.xlabel("Row within cropped image")
plt.ylabel("Mean grayscale intensity (0-255)")
plt.title("A2 - Grayscale Intensity by Row")
plt.grid(True)
plt.show()
```

```python
line_band = gray[7:13, :]
line_profile = line_band.mean(axis=0)
plt.figure(figsize=(14,4))
plt.plot(line_profile, linewidth=1)
plt.xlabel("Horizontal position within crop (pixels)")
plt.ylabel("MEan grayscale intensity (0-255)")
plt.title("A2 - Line Intensity Along Wall Outline")
plt.grid(True)
plt.show()
```

```python
processed = gray.copy()
mask = (gray >= 165) & (gray <= 190)
processed[mask] = 140
fig, axes = plt.subplots(2, 1, figsize=(14,4))
axes[0].imshow(gray, cmap="gray", vmin=0, vmax=255)
axes[0].set_title("Original Grayscale")
axes[1].imshow(processed, cmap="gray", vmin=0, vmax=255)
axes[1].set_title("Itensity Slicing 165-190 to 140")

for ax in axes:
  ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
processed_local = gray.copy()
band = processed_local[7:13, :]
mask = (band >= 165) & (band <= 190)
band[mask] = 140
fig, axes = plt.subplots(2,1,figsize=(13,4))
axes[0].imshow(gray, cmap="gray", vmin=0, vmax=255)
axes[0].set_title("Original")
axes[1].imshow(processed_local, cmap="gray", vmin=0, vmax=255)
axes[1].set_title("Spatially Restricted Intensity Slicing")

for ax in axes:
  ax.axis("off")

plt.tight_layout()
plt.show()
```

####Experiment A2 Conclusion
Intensity slicing resulted in darkened wall fill and unexpected artifacts including white splotches and a phantom horizontal stripe. This is not what was expected. In future experiment, consider rotating the line first or use a "line aware method".
